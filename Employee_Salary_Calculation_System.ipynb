```python
employees = []


def add_employee():
    print("\n--- Add Employee ---")

    employee_id = input("Enter Employee ID: ")
    name = input("Enter Employee Name: ")

    basic = float(input("Enter Basic Salary: "))
    hra = float(input("Enter HRA: "))
    da = float(input("Enter DA: "))
    other_allowances = float(input("Enter Other Allowances: "))
    deductions = float(input("Enter Deductions: "))

    gross_salary = basic + hra + da + other_allowances
    net_salary = gross_salary - deductions

    employee = {
        "Employee ID": employee_id,
        "Name": name,
        "Basic Salary": basic,
        "HRA": hra,
        "DA": da,
        "Other Allowances": other_allowances,
        "Deductions": deductions,
        "Gross Salary": gross_salary,
        "Net Salary": net_salary
    }

    employees.append(employee)

    print("\nEmployee added successfully!")
    print("Gross Salary:", gross_salary)
    print("Net Salary:", net_salary)


def display_employees():
    print("\n--- Employee Details ---")

    if not employees:
        print("No employee records found.")
        return

    for employee in employees:
        print("\nEmployee ID:", employee["Employee ID"])
        print("Name:", employee["Name"])
        print("Basic Salary:", employee["Basic Salary"])
        print("HRA:", employee["HRA"])
        print("DA:", employee["DA"])
        print("Other Allowances:", employee["Other Allowances"])
        print("Deductions:", employee["Deductions"])
        print("Gross Salary:", employee["Gross Salary"])
        print("Net Salary:", employee["Net Salary"])


def search_employee():
    employee_id = input("Enter Employee ID to search: ")

    for employee in employees:
        if employee["Employee ID"] == employee_id:
            print("\nEmployee Found!")
            print("Employee ID:", employee["Employee ID"])
            print("Name:", employee["Name"])
            print("Gross Salary:", employee["Gross Salary"])
            print("Net Salary:", employee["Net Salary"])
            return

    print("Employee not found.")


while True:
    print("\n===== Employee Salary Calculation System =====")
    print("1. Add Employee")
    print("2. Display All Employees")
    print("3. Search Employee")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_employee()

    elif choice == "2":
        display_employees()

    elif choice == "3":
        search_employee()

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice. Please try again.")
```
